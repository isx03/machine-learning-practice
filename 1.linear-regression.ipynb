# Linear Regression with library


```python
# import all libraries
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.preprocessing import StandardScaler, RobustScaler
```

```python
# see summaryse of dataset
df = pd.read_csv('./data/happiness.csv')
df.head()
```

```python
df.info()
```

```python
x = df['income'].values
y = df['happiness'].values
plt.plot(x, y, '.')

# labels
plt.title('Relación: Income vs Hapinness', fontsize=14, fontweight='bold')
plt.xlabel('Income', fontsize=12)
plt.ylabel('Hapinness.', fontsize=12)

# cuadricula
plt.grid(True, alpha=0.3, linestyle='-')
plt.show()
```

```python
# split data for traint and test
x = df[['income']].values
y = df['happiness'].values

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.35, random_state=42)

linea_regression = LinearRegression()
linea_regression.fit(x_train, y_train)

w = linea_regression.coef_[0] # pendiente
b = linea_regression.intercept_ # intercepto

print(f"w: {w}")
print(f"b: {b}")
```

```python
# predictions
y_pred = w * x + b

# Graficar
plt.figure(figsize=(10, 6))
plt.scatter(x, y, alpha=0.6, label='Real Data')
plt.plot(x, y_pred, 'r-', linewidth=2, label='Regression Line')
plt.xlabel('Income')
plt.ylabel('Happiness')
plt.title('Linear Regression: Income vs Happiness')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

# predict
y_predicted = linea_regression.predict(x_test)

# MAE
mae = mean_absolute_error(y_test, y_predicted)
print(f"MAE: {mae:.1f}")

# MSE
mse = mean_squared_error(y_test, y_predicted)
print(f"MSE: {mse:.1f}")

# RMSE
rmse = np.sqrt(mse)
print(f"RMSE: {rmse:.1f}")

# R2
r2 = r2_score(y_test, y_predicted)
print(f"R2: {r2:.3f}")
```

# Prediction with scaled data

```python
# withscaler

scaler = StandardScaler()
df_scaled = pd.DataFrame(scaler.fit_transform(df), columns=df.columns)

plt.plot(df_scaled['income'].values, df_scaled['happiness'].values, '.')
plt.grid(True, alpha=0.3)
plt.show()
```

```python
x_scaled = df[['income']].values
y_scaled = df['happiness'].values

x_train_scaled, x_test_scaled, y_train_scaled, y_test_scaled = train_test_split(x_scaled, y_scaled, test_size=0.35, random_state=42)

linea_regression = LinearRegression()
linea_regression.fit(x_train_scaled, y_train_scaled)

w = linea_regression.coef_[0] # pendiente
b = linea_regression.intercept_ # intercepto

print(f"w: {w}")
print(f"b: {b}")
```

```python
# predictions
y_pred = w * x + b

# Graficar
plt.figure(figsize=(10, 6))
plt.scatter(x, y, alpha=0.6, label='Real Data')
plt.plot(x, y_pred, 'r-', linewidth=2, label='Regression Line')
plt.xlabel('Income')
plt.ylabel('Happiness')
plt.title('Linear Regression: Income vs Happiness')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

# predict
y_predicted = linea_regression.predict(x_test_scaled)

# MAE
mae = mean_absolute_error(y_test_scaled, y_predicted)
print(f"MAE: {mae:.1f}")

# MSE
mse = mean_squared_error(y_test_scaled, y_predicted)
print(f"MSE: {mse:.1f}")

# RMSE
rmse = np.sqrt(mse)
print(f"RMSE: {rmse:.1f}")

# R2
r2 = r2_score(y_test_scaled, y_predicted)
print(f"R2: {r2:.3f}")
```
